Write a Python function that takes a list of words and return the longest word and the length of the longest one

(Longest word: Exercises
Length of the longest word: 9)

````python
words = ["Hello", "everyone", "good", "afternoon"]
def longest_word(words):
    longest = ""
    for w in words:
        if len(w) > len(longest):
            longest = w
    return longest, len(longest)

longest, length = longest_word(words)

print(f"Longest word is: {longest}")
print(f"Length of the longest word is: {length}")
````

A prime number is a number that is greater than 1 and has only two factors which are 1 and itself. The first few prime numbers are 2, 3, 5, 7, 11, 13, 17, 19, 23 and 29.

Given a number N. Determine whether N is prime or not.

Note: Solve this problem using function.

Input
First line will contain a number T (1 ≤ T ≤ 103) number of test cases.

Next T lines will contain a number N (1 ≤ N ≤ 109).

Output
Print "YES" if the Nth number is prime otherwise, print "NO".

Example

Input

3

2

4

8

Output

YES

NO

NO

````python
#my way

def is_prime(n):
    if n < 2:
        return False
    
    for i in range(2, n):
        if n % i == 0:
            return False
        return True
        
T = int(input("Number of test cases"))

for _ in range(T):
    N = int(input())
    if is_prime(N):
        print(f"{N} :Yes")
    else:
        print(f"{N} :No")
````

````python
#chatgpt way
def is_prime(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    # Check only odd numbers up to sqrt(n)
    i = 3
    while i * i <= n:
        if n % i == 0:
            return False
        i += 2

    return True


# Handling test cases
T = int(input())
for _ in range(T):
    N = int(input())
    if is_prime(N):
        print("YES")
    else:
        print("NO")
````

Write a function called fibonacci(n) that:

Takes an integer N (1 ≤ N ≤ 45) as a parameter.

Returns a list containing the first N numbers of the Fibonacci sequence.

Use the following definition:

fib(1) = 0

fib(2) = 1

fib(n) = fib(n - 1) + fib(n - 2)

After defining the function, read an integer N from the user and print the returned Fibonacci sequence numbers separated by spaces.

🔹 Example

Input

7

Output

0 1 1 2 3 5 8

Note
For more information visit Fibonacci: https://www.mathsisfun.com/numbers/fibonacci-sequence.html.

````python
def fibonacci(n):
    if n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    
    fib_sequence = [0, 1]
    for i in range(2, n):
        nxt_num = fib_sequence[i -1] + fib_sequence[i - 2]
        fib_sequence.append(nxt_num)
    return fib_sequence

N = int(input())

print(*fibonacci(N))
````

```
Note
Don't print any extra space after '*'.
```

````python
def pyramid(n, current=1):
    if current > n:
        return
    
    stars = 2 * current - 1
    spaces = n - current
    
    print(" " * spaces + "*" * stars)
    
    pyramid(n, current + 1)

N = int(input())
pyramid(N)
````

Write a function that takes an integer number as input and returns a list containing each digit of the number separately.

````python
def split_digits(number):
    return [int(d) for d in str(number)]

print(split_digits(12345))
````
